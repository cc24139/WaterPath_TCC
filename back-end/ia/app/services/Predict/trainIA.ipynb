```python
data = pd.read_csv(os.getcwd() + "/station/resultphyschem.csv",low_memory=False)

#data["CharacteristicName"].value_counts()
data["ActivityMediaName"].value_counts()
```

```python
parameters = data["CharacteristicName"].value_counts().head(10).index.tolist()
data = data[data["CharacteristicName"].isin(parameters)]
data["ResultMeasureValue"] = pd.to_numeric(data["ResultMeasureValue"], errors='coerce')
table = data.pivot_table(index=['ActivityMediaName',"MonitoringLocationIdentifier"], columns='CharacteristicName', values='ResultMeasureValue', aggfunc='mean').reset_index()

table
```
