## KNN
Utiliza a informação dos k vizinhos mais proximos para determinar a característica do dado de entrada

```python
from sklearn.neighbors import KNeighborsClassifier
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
import plotly.express as px
from sklearn.model_selection import train_test_split as sklearn
```

```python
x,y = load_iris(return_X_y=True)
X_train , x_Test , y_train , y_test = sklearn(x,y,test_size=0.3)
```

```python
numK = 10
clf = KNeighborsClassifier(n_neighbors=numK)
clf.fit(X_train,y_train)
```

```python
result = clf.predict(x_Test)
iris = load_iris()
target_name = iris.target_names
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(confusion_matrix(y_test,result))
print(classification_report(y_test,result,target_names = target_name))
```
