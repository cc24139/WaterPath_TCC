```python
from PIL import Image
import os

# pasta com imagens originais
input_dir = "datasetV2/all_images/"
# pasta de saída
output_dir = "datasetV2/processed/"

# configurações
TARGET_SIZE = (800, 800)
FORMAT = "JPEG"  # pode ser PNG também

os.makedirs(output_dir, exist_ok=True)

for filename in os.listdir(input_dir):
    filepath = os.path.join(input_dir, filename)

    try:
        with Image.open(filepath) as img:
            # converter para RGB (remove alpha, padroniza cor)
            img = img.convert("RGB")

            # redimensionar
            img = img.resize(TARGET_SIZE)

            # nome novo (sempre .jpg)
            new_name = os.path.splitext(filename)[0] + ".jpg"
            output_path = os.path.join(output_dir, new_name)

            # salvar
            img.save(output_path, FORMAT, quality=95)

            print(f"Convertido: {filename}")

    except Exception as e:
        print(f"Erro em {filename}: {e}")
```
